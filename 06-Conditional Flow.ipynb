# if - elif - else

```python
grade = None
score = float(input("enter student score"))
if score < 50 and score >=0:
    grade = "F"
elif score < 80 and score >= 70:
    grade = "C"
elif score > 90 and score <= 100:
    grade = "A"
elif score > 80 and score <=90:
    grade = "B"

print("Your grade is " + grade + "!!!")
print(f"Your grade is {grade}!!!")
print(f"Your grade is grade!!!")
```

# Nested if-else

```python
is_login = True
is_admin = False

if is_login:
    if is_admin:
        print("Go to admin dashboard")
    else:
        print("Go to user panel")
else:
    print("Please login first!")
```

```python
if is_login and is_admin:
    print("Go to admin dashboard")
elif is_login and not is_admin:
    print("Goto user panel")
else:
    print("Please login first")
```
