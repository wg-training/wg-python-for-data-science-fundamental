```python
data = { "id": 1, "amount": "1000"}
try:
    exchange = int(data["amount"]) * 4100
except (ValueError, KeyError):
    print("Amount is invalid")
    exchange = None

if exchange is not None:
    print(f"Exchanged amount: {exchange}")
```

```python
data = { "id": 1}
try:
    exchange = int(data["amount"]) * 4100
except ValueError:
    print("Amount value is invalid")
    exchange = None
except KeyError:
    print("Data has no amount")

if exchange is not None:
    print(f"Exchanged amount: {exchange}")
```

```python
import traceback
data = { "id": 1, "amount": "a"}
try:
    exchange = int(data["amount"]) * 4100
except Exception as e:
    print("Amount value is invalid")
    # print(type(e))
    # print(type(e).__name__)
    print(traceback.print_exc())
    exchange = None

if exchange is not None:
    print(f"Exchanged amount: {exchange}")
```
