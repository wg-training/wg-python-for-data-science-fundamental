# Arithmetic Operator

```python
a = 17
b = 5

print(a + b)  # want 22      (add)
print(a - b)  # want 12      (subtract)
print(a * b)  # want 85      (multiply)
print(a / b)  # want 3.4     (divide include floating)
print(a // b)  # want 3       (divide whole number only)
print(a % b)  # want 2       (remainder)
print(a ** b)  # want 1419857 (a power b)
```

```python
print(17 // 5)
print(17 % 5)
```

```python
name = "Python"
line = "="

greeting = "Hi " + name + "!"
edge = line * 20

print(edge)
print(greeting)
print(edge)
```
