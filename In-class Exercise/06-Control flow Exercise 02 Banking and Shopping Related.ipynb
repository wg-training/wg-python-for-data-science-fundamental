### Account Tier

```python
balance = float(input("Enter balance: "))

# TODO: classify to Platinum / Gold / Silver / Basic
# Platinum >=10000, Gold >=5000, Silver >=1000, else Basic
if balance >= 10000:
    print("Platinum")
elif balance >= 1000 and balance<5000:
    print("Siliver")
elif balance >= 5000 and balance<10000:
    print("Gold")
else:
    print("Basic")
    
```
