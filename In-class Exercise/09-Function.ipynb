```python
def format_full_name(first_name, last_name, middle_name=""):
    # your code here
    if middle_name == "":
        return f"{first_name} {last_name}"
    return f"{first_name} {middle_name} {last_name}"

name = format_full_name("Ana","Lee")
print(name)
name = format_full_name("Ana","Lee","J")
print(name)
```

```python
def format_full_name(first_name, last_name, middle_name=""):
    # your code here
    middle_name_fmt = middle_name + " " if middle_name !="" else ""
    return f"{first_name} {middle_name_fmt}{last_name}"

name = format_full_name("Ana","Lee")
print(name)
name = format_full_name("Ana","Lee","J")
print(name)
```

## Exercise 10

```python
def clean_text(text):
    # your code here
    return text.strip().lower()

def count_words(text):
    # your code here
    #return list(filter(lambda t: t!="", text.split(" ") )
    words = []
    for t in text.split(" "):
        if t != "":
         words.append(t)
    return len(words)

def process_document(word):
    # your code here
    txt = clean_text(word)
    print("Cleaned text -> ", txt)
    txt = count_words(txt)
    print("Count text -> ", txt)


processed_doc = process_document("    HellO PYthon    PROGRAMer   ")
```
