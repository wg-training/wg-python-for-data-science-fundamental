6 Multiplication Table

```python
count = 1
while count <=10:
    print(f"5 X {count} = {5*count}")
    #count = count + 1
    count += 1
print("Done!")
```

```python
word = "programming"
vowels = "aeiou"
wi = 0
count_vowel = 0

while wi < len(word):

    vi = 0
    while vi < len(vowels):
        if word[wi] == vowels[vi]:
            count_vowel = count_vowel + 1
        vi = vi +1 


    wi = wi +1
print(f"Total vowels: {count_vowel}")
```

```python
word = "programming"
vowels = ["a", "e", "i", "o", "u"]
word_index = 0
print(word[word_index])
count = 0
while word_index < len(word):
    count += vowels.count(word[word_index])
    word_index = word_index + 1
print(f"Total Vowels: {count}")
```

```python
word = "programming"
vowels = 'aeiou'
vi = 0
count = 0
while vi < len(vowels):
    count += word.count(vowels[vi])
    vi = vi + 1
    
print(f"Total Vowels: {count}")
```

8 Find largest number

```python
numbers = [-1001, 58, 2, 99, 43]
number_index = 0
max_number = numbers[0]
# print(max_number)
while number_index < len(numbers):
    if numbers[number_index] > max_number:
        max_number = numbers[number_index]
    number_index = number_index + 1

print(f"Max number: {max_number}")
```
