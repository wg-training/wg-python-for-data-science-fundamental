Exercise 3: 

Sum of Numbers


```python
# total: 0 + 1
# step 2: total of step1(0+1) + 2
# step 3: total of step 2(0+1+2) + 3
# step 4: total of step 3(0+1+2+3) + 4
prev_total = 0
formulat = ""
for n in range(1, 101):
    prev_total = (prev_total + n)
    formulat = f"{formulat}+{n}"
    # print("Cumulative total="+ str(formulat))
print(prev_total)
```
